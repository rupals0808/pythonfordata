## 1. Python's Role in Data Engineering

Python is the go-to language for data engineering because:

- It acts as a **glue language** — connects databases, APIs, cloud services, file systems
- Has a huge ecosystem — pandas, PySpark, Airflow, SQLAlchemy, requests
- ETL pipelines are very natural to write in Python
- All major cloud platforms support Python natively

In a typical data pipeline:

```
Data Sources (APIs, CSVs, DBs)
        ↓
   EXTRACT  ← Python (requests, file I/O, connectors)
        ↓
  TRANSFORM  ← Python (pandas, PySpark)
        ↓
     LOAD    ← Python (SQLAlchemy, cloud SDKs)
        ↓
Data Warehouse (BigQuery, Snowflake, etc.)
```

As a data engineer, you don't just analyze data — you build the **pipelines** that move and transform it.

## 2. Variables & Data Types

A variable is just a name that points to a value.

Python has 4 basic data types you'll use all the time:

| Type | What it is | Example | Used for |
|------|-----------|---------|----------|
| `int` | Whole numbers | `42` | Row counts, IDs |
| `float` | Decimal numbers | `3.14` | Prices, metrics |
| `str` | Text | `"hello"` | File paths, column names |
| `bool` | True/False | `True` | Flags, validation |

````python
# Variables and their types

rows_processed = 15000          # int
avg_time = 2.45                 # float
source_table = "raw_sales"      # str
is_active = True                # bool

print(rows_processed, type(rows_processed))
print(avg_time, type(avg_time))
print(source_table, type(source_table))
print(is_active, type(is_active))
````

````python
# Type conversion - very common when reading CSV data

value = "42"                    # this is a string, not a number
print(type(value))

value_int = int(value)          # string -> int
value_float = float(value)      # string -> float
back_to_str = str(value_int)    # int -> string

print(value_int, type(value_int))
print(value_float, type(value_float))
print(back_to_str, type(back_to_str))

# bool conversion
print(bool(1))       # True
print(bool(0))       # False
print(bool(""))      # False (empty string)
print(bool("hello")) # True  (non-empty string)
````

### 3.3 Sets

Unordered, **no duplicates**. Great for deduplication and comparing datasets.

````python
# Deduplication
raw_ids = [101, 102, 103, 101, 104, 102, 105]
unique_ids = set(raw_ids)
print(f"Raw: {len(raw_ids)} items, Unique: {len(unique_ids)} items")
print(unique_ids)

# Set operations - comparing columns between tables
source_cols = {"id", "name", "email", "phone", "created_at"}
target_cols = {"id", "name", "email", "address", "updated_at"}

print(f"Common:         {source_cols & target_cols}")        # intersection
print(f"Only in source: {source_cols - target_cols}")        # difference
print(f"Only in target: {target_cols - source_cols}")        # difference
print(f"All combined:   {source_cols | target_cols}")        # union

# Fast lookup
print(f"'email' in source? {'email' in source_cols}")  # True
````

## 4. Basic Operators

| Category | Operators | Example use |
|----------|-----------|------------|
| Arithmetic | `+  -  *  /  //  %  **` | Batch size calc, metrics |
| Comparison | `==  !=  >  <  >=  <=` | Data validation |
| Logical | `and  or  not` | Combine conditions |
| Membership | `in  not in` | Check if value exists |

````python
# Arithmetic - batch processing calculations

total_records = 15000
batch_size = 1000

num_batches = total_records // batch_size   # floor division = 15
remaining = total_records % batch_size      # modulo (remainder) = 0
print(f"Batches: {num_batches}, Remaining: {remaining}")

time_per_batch = 2.5
total_time = num_batches * time_per_batch
print(f"Est. time: {total_time}s")

# Exponent
file_size_bytes = 1024 ** 3  # 1 GB
print(f"1 GB = {file_size_bytes} bytes")
````

````python
# Comparison + Logical - data validation

row_count = 15000

# Is the count in valid range?
is_valid = row_count >= 10000 and row_count <= 20000
print(f"Valid count? {is_valid}")

# Does data need cleaning?
has_nulls = False
has_duplicates = True
needs_cleaning = has_nulls or has_duplicates
print(f"Needs cleaning? {needs_cleaning}")

# Membership check
valid_statuses = ["active", "pending", "completed"]
print(f"'active' valid? {'active' in valid_statuses}")
print(f"'failed' valid? {'failed' in valid_statuses}")
````

### Lab 2: Simple ETL Transformation

A mini ETL pipeline: Extract raw data → Transform (clean, filter, enrich) → Load clean data.

````python
# Example ETL pipeline

# EXTRACT
raw_data = [
    {"name": "  alice  ", "city": "NEW YORK", "amount": "120.50", "status": "completed"},
    {"name": "BOB", "city": "  chicago ", "amount": "89.99", "status": "completed"},
    {"name": " Charlie ", "city": "los angeles", "amount": "invalid", "status": "failed"},
    {"name": "diana", "city": "HOUSTON", "amount": "250.00", "status": "completed"},
]

print(f"Extracted {len(raw_data)} records")

# TRANSFORM
clean_data = []
errors = []

for record in raw_data:
    if record["status"] != "completed":
        errors.append(record)
        continue
    
    try:
        amount = float(record["amount"])
    except ValueError:
        errors.append(record)
        continue
    
    clean_record = {
        "name": record["name"].strip().title(),
        "city": record["city"].strip().title(),
        "amount": amount,
        "tax": round(amount * 0.08, 2),
        "total": round(amount * 1.08, 2),
    }
    clean_data.append(clean_record)

# LOAD (just printing here, in real life you'd write to DB/file)
print(f"Transformed: {len(clean_data)} clean, {len(errors)} errors")
for r in clean_data:
    print(f"  {r}")

total_revenue = sum(r["total"] for r in clean_data)
print(f"Total revenue: ${total_revenue:,.2f}")
````

### Lab 3: Parse Raw Data & Structure It

Very common DE task: take unstructured text and turn it into structured data.

````python
# Example: parsing log lines

raw_logs = [
    "2024-01-15 10:23:45 | INFO | user_login | user_id=1001 | ip=192.168.1.10",
    "2024-01-15 10:24:12 | ERROR | db_connection | user_id=1002 | ip=192.168.1.15",
    "2024-01-15 10:25:30 | INFO | data_export | user_id=1001 | ip=192.168.1.10",
    "2024-01-15 10:26:01 | WARNING | slow_query | user_id=1003 | ip=10.0.0.5",
]

parsed = []
for log in raw_logs:
    parts = log.split(" | ")
    parsed.append({
        "timestamp": parts[0],
        "level": parts[1],
        "event": parts[2],
        "user_id": int(parts[3].split("=")[1]),
        "ip": parts[4].split("=")[1],
    })

for log in parsed:
    print(log)

# Quick analysis
errors = sum(1 for l in parsed if l["level"] == "ERROR")
unique_users = set(l["user_id"] for l in parsed)
print(f"\nErrors: {errors}, Unique users: {unique_users}")
````

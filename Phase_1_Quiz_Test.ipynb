## Coding Task Grading (10 points)

Run the cell below to auto-grade your `process_sales_data` function.  
It checks 10 criteria — each worth 1 point.

````python
# Coding Task Auto-Grader

def grade_coding_task():
    checks = []
    code_score = 0
    code_total = 10

    # Run the function
    try:
        result = process_sales_data("data/quiz_sales.csv")
    except Exception as e:
        print("=" * 70)
        print("  CODING TASK - AUTO-GRADER RESULTS")
        print("=" * 70)
        print(f"\n  [FAIL]  Function crashed: {type(e).__name__}: {e}")
        print(f"\n  Coding Task Score: 0/{code_total} (0%)")
        print("  Fix the error and run this cell again.")
        print(f"\n{'=' * 70}")
        return 0

    # Check 1: returns a dict
    if isinstance(result, dict):
        checks.append(("PASS", "Returns a dictionary"))
        code_score += 1
    else:
        rtype = type(result).__name__ if result is not None else "None"
        checks.append(("FAIL", f"Should return dict, got {rtype}"))
        # Can't check further if not a dict
        print("=" * 70)
        print("  CODING TASK - AUTO-GRADER RESULTS")
        print("=" * 70)
        for s, m in checks:
            print(f"\n  [{s}]  {m}")
        print(f"\n  Coding Task Score: {code_score}/{code_total} ({code_score/code_total*100:.0f}%)")
        print(f"\n{'=' * 70}")
        return code_score

    # Check 2: has all required keys
    required = {"clean_data", "error_count", "total_revenue", "by_region"}
    if required.issubset(set(result.keys())):
        checks.append(("PASS", "Has all 4 required keys"))
        code_score += 1
    else:
        missing = required - set(result.keys())
        checks.append(("FAIL", f"Missing keys: {missing}"))

    # Check 3: clean_data is a list
    if isinstance(result.get("clean_data"), list):
        checks.append(("PASS", "clean_data is a list"))
        code_score += 1
    else:
        checks.append(("FAIL", "clean_data should be a list"))

    # Check 4: correct number of clean records = 6
    n_clean = len(result.get("clean_data", []))
    if n_clean == 6:
        checks.append(("PASS", f"Clean record count: {n_clean} (expected 6)"))
        code_score += 1
    else:
        checks.append(("FAIL", f"Clean record count: {n_clean} (expected 6)"))

    # Check 5: correct error count = 4
    err = result.get("error_count", -1)
    if err == 4:
        checks.append(("PASS", f"Error count: {err} (expected 4)"))
        code_score += 1
    else:
        checks.append(("FAIL", f"Error count: {err} (expected 4)"))

    # Check 6: total_revenue close to 5421.89
    rev = result.get("total_revenue", 0)
    if isinstance(rev, (int, float)) and abs(rev - 5421.89) < 1.0:
        checks.append(("PASS", f"Total revenue: ${rev:,.2f} (expected ~$5,421.89)"))
        code_score += 1
    else:
        checks.append(("FAIL", f"Total revenue: {rev} (expected ~$5,421.89)"))

    # Check 7: by_region is a dict
    by_region = result.get("by_region", None)
    if isinstance(by_region, dict) and len(by_region) > 0:
        checks.append(("PASS", f"by_region is a dict with {len(by_region)} regions"))
        code_score += 1
    else:
        checks.append(("FAIL", "by_region should be a non-empty dict"))

    # Check 8: has 3 correct regions
    expected_regions = {"north", "south", "east"}
    if isinstance(by_region, dict) and set(by_region.keys()) == expected_regions:
        checks.append(("PASS", f"Correct regions: {sorted(by_region.keys())}"))
        code_score += 1
    else:
        got_r = set(by_region.keys()) if isinstance(by_region, dict) else set()
        checks.append(("FAIL", f"Regions: {sorted(got_r)} (expected {sorted(expected_regions)})"))

    # Check 9: clean records have 'total' field
    clean = result.get("clean_data", [])
    if clean and "total" in clean[0]:
        checks.append(("PASS", "Clean records include 'total' field"))
        code_score += 1
    else:
        checks.append(("FAIL", "Clean records should have a 'total' field (price * quantity)"))

    # Check 10: product names are cleaned (title case, stripped)
    if clean:
        p = clean[0].get("product", "")
        if p and p == p.strip() and p == p.strip().title():
            checks.append(("PASS", f"Product names cleaned (e.g. '{p}')"))
            code_score += 1
        else:
            checks.append(("FAIL", f"Product not cleaned: '{p}' (expected title case, stripped)"))
    else:
        checks.append(("FAIL", "No clean records to check product names"))

    # Print detailed results
    print("=" * 70)
    print("  CODING TASK - AUTO-GRADER RESULTS (10 checks)")
    print("=" * 70)
    print()
    for idx, (status, msg) in enumerate(checks, 1):
        print(f"  {idx:2d}. [{status}]  {msg}")

    code_pct = code_score / code_total * 100
    print(f"\n  {'_' * 60}")
    print(f"  CODING TASK SCORE:  {code_score}/{code_total}  ({code_pct:.0f}%)")

    if code_pct == 100:
        print("  Excellent! Your pipeline handles all cases correctly.")
    elif code_pct >= 70:
        print("  Good work! Fix the FAIL items above.")
    elif code_pct >= 40:
        print("  Partial implementation. Review failed checks and try again.")
    else:
        print("  Needs more work. Re-read the task requirements.")

    # Combined overall score
    combined = mcq_correct + code_score
    combined_total = 30
    combined_pct = combined / combined_total * 100

    print(f"\n{'=' * 70}")
    print(f"  OVERALL PHASE 1 SCORE")
    print(f"{'=' * 70}")
    print(f"\n  MCQ:          {mcq_correct}/20  ({mcq_correct/20*100:.0f}%)")
    print(f"  Coding Task:  {code_score}/10  ({code_pct:.0f}%)")
    print(f"  {'_' * 35}")
    print(f"  TOTAL:        {combined}/{combined_total}  ({combined_pct:.0f}%)\n")

    if combined_pct >= 80:
        print("  You're ready for Phase 2!")
    elif combined_pct >= 60:
        print("  Passing. Review weak areas before starting Phase 2.")
    else:
        print("  Not passing yet. Revisit the sessions and try again.")

    print(f"\n{'=' * 70}")
    return code_score


# Run grader (make sure you ran the MCQ grading cell first!)
try:
    mcq_correct
except NameError:
    print("Run the MCQ grading cell first, then run this cell.")
else:
    grade_coding_task()
````
